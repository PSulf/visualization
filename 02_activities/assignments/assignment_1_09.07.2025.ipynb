```python
import numpy as np 
import matplotlib.pyplot as plt
import pandas as pd 
import scipy
import PIL
import requests
# first make our sample data
x = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])

# define our figure and axes (just one this time)
fig, ax = plt.subplots(figsize=(7, 3))

# now call both bar and plot elements to the same axes (ax)
ax.bar(x, y1,
color = "indigo")
ax.plot(x, y2,
color = "red")

```

```python
y2_sd = np.std(y2)
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "brown")

ax.errorbar(x, #our x values
        y2, #our y values
        yerr = y2_sd,
        fmt = "none")
```

```python
# customizing error bars
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2,
        color = "red")
ax.errorbar(x,
        y2,
        yerr = y2_sd,
        fmt = "none",
        ecolor= "indigo",
        elinewidth= 4,
        capsize = 6,
        capthick= 4 )
```

```python
ax.errorbar(x,
        y2,
        yerr = y2_sd,
        fmt = "none",
        ecolor = "indigo",
        elinewidth = 4,
        capsize = 6,
        capthick = 4,
        errorevery= 2 )
```

```python
import matplotlib.pyplot as plt
from PIL import Image 
import requests 
from io import BytesIO 

# Load image from the internet
response = requests.get('https://upload.wikimedia.org/wikipedia/en/c/cb/Monkey_D_Luffy.png')
image_file = BytesIO(response.content)
image = Image.open(image_file)

# data for plotting
x = [1, 2, 3, 4, 5]
y2 = [1, 4, 9, 16, 25]

#  unpacking of fig and ax
fig, ax = plt.subplots(figsize=(7, 3))

# Plot data
ax.plot(x, y2, color='red')

# Add image as a subplot/inset
ax_image = fig.add_axes([0.1, 0.11, 0.15, 0.35])
ax_image.imshow(image)
ax_image.axis('off')

# Display the final figure
plt.show()
```
