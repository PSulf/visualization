```python
# Session 02 – Getting Started with matplotlib

#In this session, we learned:
#- How to import `matplotlib.pyplot` and create basic line plots
#- How to add labels, titles, and save figures
#- The difference between scripts and interactive plotting

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests

np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0, 100,50)

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.bar(x,y)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
```

```python
#histogram
fig, ax = plt.subplots(figsize=(5, 3))
ax.hist(y)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.hist(y)

ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')

font1 = {'family':'sans-serif','color':'blue','size':20}
font2 = {'family':'monospace','color':'green','size':14}
```

```python
# modify
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax.set_title('Total growth over time', fontdict = font1)
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Years since start', fontdict = font2)
```

```python
# moving labels and titles
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax.set_title('Total growth over time', fontdict = font1, loc = 'left')
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Years since start', fontdict = font2)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))

ax.plot(x,y,marker='*',
        markersize = 12,              
        color = '#7425b9',
        linestyle = '--',
        linewidth = 2,
        markeredgecolor = '#fa9359',  
        markerfacecolor = '#fa9359' ) 
#grid lines
ax.grid(axis = 'y')
fig.show()
```

```python
# Session 04 – Choosing the Right Visualization

## Key Takeaways:
#- The **type of data** (categorical vs. continuous) determines the kind of chart to use
#- Charts must reflect the **purpose**: comparison, distribution, trend, or relationship
#- Avoid misleading visuals (e.g., improper y-axis scaling)
#- Visual accessibility and clarity matter more than aesthetics

## Common Chart Types:
#| Purpose        | Visualization Types                      |
#|----------------|------------------------------------------|
#| Comparison     | Bar chart, Column chart, Grouped bar     |
#| Distribution   | Histogram, Boxplot, Violin plot          |
#| Trend over time| Line plot, Area chart                    |
#| Relationship   | Scatter plot, Bubble plot                |
```
